```python
import numpy as np
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

```

```python
data = pd.read_csv("heart_disease_data.csv")
```

```python
X = data.drop(columns="target",axis=1)
Y= data["target"]
```

```python
X
```

```python
Y
```

```python
X_train,X_test,Y_train,Y_test= train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
```

```python
X.shape,X_train.shape,X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler

# StandardScaler nesnesini oluştur
scaler = StandardScaler()

# X_train ve X_test verilerini ölçeklendir
X_train_scaled = scaler.fit_transform(X_train)  # Eğitim verisini ölçeklendir
X_test_scaled = scaler.transform(X_test)  # Test verisini aynı ölçeğe getir
```

```python
model = LogisticRegression(solver="saga",max_iter=5000)
```

```python
model.fit(X_train_scaled,Y_train)
```

```python
X_train_prediction= model.predict(X_train_scaled)
training_data_accuracy= accuracy_score(X_train_prediction,Y_train)
```

```python
training_data_accuracy
```

```python
X_test_prediction= model.predict(X_test_scaled)
test_data_accuracy= accuracy_score(X_test_prediction,Y_test)
```

```python
test_data_accuracy
```
